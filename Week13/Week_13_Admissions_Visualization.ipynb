First calculate application frequencies with Pandas. Recreate the same comparison using each visualization library.</div></div>

```python
# Step 1: Calculate applications per college
college_counts = df['College'].value_counts()
college_df = college_counts.rename_axis('College').reset_index(name='Applications')
display(college_df)

# Step 2: Identify the most popular college
print('Most applications:', college_counts.idxmax(), '-', college_counts.max())
```

```python
# Step 1: Import libraries
import numpy as np

# Step 2: Define class boundaries (100-point intervals)
bins = np.arange(200, 1701, 100)

# Step 2: Calculate frequencies and interval boundaries
frequency, intervals = np.histogram(df['SAT_ACT'], bins=bins)
frequency_table = pd.DataFrame({
    'Score Range': [f'{int(intervals[i])}–{int(intervals[i+1]-1)}' for i in range(len(frequency))],
    'Frequency': frequency
})
display(frequency_table)
print('Most common score range:', frequency_table.loc[frequency_table['Frequency'].idxmax(), 'Score Range'])
print('Mean SAT_ACT:', round(df['SAT_ACT'].mean(), 2))
```

Follow the professor’s scatter plot demonstrations by adding a linear trendline in all three libraries.</div></div>

```python
# Step 1: Import libraries
import numpy as np

# Step 2: Calculate Pearson correlation
correlation = df['SAT_ACT'].corr(df['Extracurricular_Hours'])
print('Pearson correlation:', round(correlation, 5))

# Step 2: Calculate a linear trendline using NumPy
m, b = np.polyfit(df['SAT_ACT'], df['Extracurricular_Hours'], 1)
print('Trendline slope:', round(m, 6))
```

<div style="font-family:'Times New Roman',serif;text-align:center;background-color:#F5F8FC;padding:15px;border-radius:10px"><h2 style="color:#2563EB;font-size:28px">Scatter Plot Using Matplotlib</h2></div>

<p style="font-family:'Times New Roman',serif;font-size:17px"><b>Task:</b> Visualize SAT_ACT scores and extracurricular hours with a linear trendline.</p>

```python
# Step 1: Import libraries
import matplotlib.pyplot as plt
import numpy as np

# Step 2: Set figure size
plt.figure(figsize=(8, 5))

# Step 2: Create scatter plot
plt.scatter(df['SAT_ACT'], df['Extracurricular_Hours'],
            color='blue', s=10, alpha=0.3)

# Step 3: Draw linear trendline
x_line = np.array([df['SAT_ACT'].min(), df['SAT_ACT'].max()])
plt.plot(x_line, m * x_line + b, color='red')

# Step 4: Add labels and display
plt.title('SAT_ACT vs. Extracurricular Hours — Matplotlib')
plt.xlabel('SAT_ACT Score')
plt.ylabel('Extracurricular Hours per Week')
plt.tight_layout()
plt.show()
```

<div style="font-family:'Times New Roman',serif;background:linear-gradient(110deg,#EAF2FF,#E8F7F1);border:1px solid #C7DDED;border-radius:10px;padding:15px 20px;margin:12px 0"><h3 style="color:#1D4ED8;margin:4px 0 8px">Task 04: Data-Driven Insights and Business Recommendations</h3><div style="font-size:15px;line-height:1.55">Use admission and enrollment outcomes to supplement the three required visualizations. Check data consistency before interpreting enrollment yield.</div></div>

```python
# Step 1: Calculate admission and enrollment measures by college
summary = df.groupby('College').agg(
    Applications=('Applicant', 'count'),
    Admitted=('Admitted', lambda s: (s == 'Yes').sum()),
    Enrolled=('Enrolled', lambda s: (s == 'Yes').sum())
)
summary['Admission Rate (%)'] = (summary['Admitted'] / summary['Applications'] * 100).round(2)

# Step 2: Calculate enrollment yield only among admitted applicants
admitted_and_enrolled = df.loc[(df['Admitted'] == 'Yes') & (df['Enrolled'] == 'Yes')].groupby('College').size()
summary['Admitted Who Enrolled'] = admitted_and_enrolled
summary['Yield Among Admitted (%)'] = (summary['Admitted Who Enrolled'] / summary['Admitted'] * 100).round(2)
display(summary)

# Step 3: Identify inconsistent records
inconsistent = df[(df['Admitted'] == 'No') & (df['Enrolled'] == 'Yes')]
print('Enrolled but not admitted:', len(inconsistent))
```

**Finding 1 — Application demand differs by college.** Arts & Letters has the most applications (6,964) but an admission rate of approximately 25%. **Recommendation:** Review admissions processing capacity and class availability for Arts & Letters; use applicant-volume forecasts to plan staffing and resources.

**Finding 2 — Enrollment conversion differs by college.** Business & Economics has a comparatively strong admitted-student enrollment yield. **Recommendation:** Examine its outreach and accepted-student communication practices, and test effective approaches in the other colleges. Calculate yield from applicants who were *both* admitted and enrolled.

**Finding 3 — Test scores and extracurricular hours show no meaningful linear association.** The correlation is approximately 0.002. **Recommendation:** Treat these as separate dimensions of applicant evaluation rather than assuming one predicts the other.

**Data quality observation:** Some records indicate enrollment even though admission is marked “No.” Verify these records before final enrollment forecasts or official reporting.
